# Q1. Explain the roles of the Driver, Cluster Manager, and Executor in a Spark application

## Driver

The Driver is the main program of a Spark application.

It creates the Spark Session.

It divides the work into small tasks.

It sends these tasks to the Executors.

It also collects the final result.


## Cluster Manager

The Cluster Manager manages all available resources.

It provides CPU and memory to the Spark application.

It starts Executors when the Driver requests resources.


## Executor

Executors run the tasks assigned by the Driver.

They process the data.

After completing the task, they send the result back to the Driver.

### Observation

The Driver controls the application.

The Cluster Manager allocates resources.

The Executors perform the actual data processing.

# Q2. How does Spark's Lazy Evaluation improve performance?

Spark uses Lazy Evaluation.

It does not execute transformations immediately.

Instead, Spark records all transformations.

Execution starts only when an Action such as show(), count(), collect(), or write() is called.

This helps Spark optimize the execution plan and reduces unnecessary work.

As a result, Spark processes large datasets more efficiently.

```python
lazy_df = movies.filter(
    col("release_year")>2020
).select(
    "title",
    "release_year",
    "vote_average"
)

print("Transformation Created")
```

```python
lazy_df.show(10,truncate=False)
```

### Observation

The filter() and select() operations are transformations.

Spark did not execute them immediately.

Execution started only after the show() action was called.

# Q4. Difference Between CSV and Parquet

| CSV | Parquet |
|------|----------|
| Row-based storage | Column-based storage |
| Larger file size | Smaller file size |
| Slower reading | Faster reading |
| No compression | Compressed |
| No schema information | Stores schema |
| Better for sharing | Better for analytics |

Parquet is faster because Spark reads only the required columns instead of the complete file.

### Observation

Parquet provides better performance than CSV.

It uses less storage space and reads data faster.

# Q7. Lineage Graph (DAG)

Spark keeps track of every transformation performed on a DataFrame.

These transformations are stored in a Directed Acyclic Graph (DAG), also called the Lineage Graph.

If a worker node fails, Spark uses this graph to recompute only the missing data instead of processing the entire dataset again.

This provides fault tolerance and makes Spark reliable.

```python
dag_df = movies.filter(
    col("release_year") > 2020
).select(
    "title",
    "release_year",
    "vote_average"
)

dag_df.explain(True)
```

### Observation

The explain(True) function displays Spark's execution plan.

Spark uses this plan to recover lost partitions if a worker node fails.

# Q9. Predicate Pushdown

Predicate Pushdown is an optimization technique used by Spark while reading Parquet files.

Instead of loading all rows into memory, Spark sends the filter condition to the storage layer.

Only the required rows are read.

This reduces disk I/O, saves memory, and improves performance.

```python
parquet_movies = movies

predicate_df = parquet_movies.filter(
    col("release_year") > 2022
)

predicate_df.explain(True)
```

### Observation

Spark applied the filter before processing the complete dataset.

This reduced unnecessary data scanning.

# Q11. Difference Between Transformations and Actions

Transformations and Actions are two important operations in Apache Spark.

## Transformations

A Transformation creates a new DataFrame from an existing DataFrame.

It does not execute immediately because Spark follows Lazy Evaluation.

Examples:
- filter()
- select()

## Actions

An Action executes all pending transformations and returns the final result.

Examples:
- show()
- count()

```python
transform_df = movies.filter(
    col("release_year") >= 2020
).select(
    "title",
    "release_year"
)

print("Transformation Created")
```

```python
print("Total Movies :", transform_df.count())

transform_df.show(5, truncate=False)
```

### Observation

The filter() and select() functions are transformations.

The count() and show() functions are actions.

Spark executed the transformations only after the action was called.

# Q13. Difference Between Client Mode and Cluster Mode

## Client Mode

In Client Mode, the Driver program runs on the local machine.

If the client machine is disconnected, the Spark application may stop.

Client Mode is mainly used for testing and development.

## Cluster Mode

In Cluster Mode, the Driver program runs inside the cluster.

The application continues running even if the client disconnects.

Cluster Mode is mainly used in production environments.

### Observation

Client Mode is suitable for development.

Cluster Mode is more reliable for large applications because the Driver runs inside the cluster.

# Q15. Why is show(5) safer than collect()?

The show(5) function displays only the first five rows of a DataFrame.

The collect() function loads all rows into the Driver's memory.

For very large datasets, collect() may consume a large amount of memory and can slow down or even crash the application.

Therefore, show(5) is the safer option for exploring large datasets.

```python
movies.show(5, truncate=False)

print("Total Movies :", movies.count())
```

### Observation

The show(5) function displayed only a few records without loading the entire dataset into memory.

This makes it suitable for exploring large datasets.
